# Ceate bronze layer

```python
spark.sql("CREATE SCHEMA IF NOT EXISTS workspace2.bronze")
```

# Ingest the tables to bronze 

```python
## Define the ingestion function

def ingest_to_bronze(table_name, file_path):

    path = f"/Volumes/workspace2/dataflow/raw_files/{file_path}"

    df = (
        spark.read
        .format("csv")
        .option("header", "true")
        .option("inferSchema", "true")
        .option("multiLine", "true")
        .load(path)
        )

    df.write.mode("overwrite").saveAsTable(f"workspace2.bronze.{table_name}")

    print(f"Loaded: {table_name}")
    print(f"Rows: {df.count()}")
```

```python
## define the tables

tables = {
    "customers": "cust_info.csv",
    "products": "prd_info.csv",
    "sales": "sales_details.csv",
    "cust_az": "CUST_AZ12.csv",
    "loc_a1": "LOC_A101.csv",
    "px_cat": "PX_CAT_G1V2.csv",
}

## call the function for ingestion

for table_name, file_path in tables.items():
    ingest_to_bronze(table_name, file_path)
```
