# INIT

```python
from pyspark.sql.functions import col, trim, abs, current_date
```

```python
## Create tables dataframe  

customer_silver = spark.table("workspace2.silver.customer")

products_silver = spark.table("workspace2.silver.products")

sales_silver = spark.table("workspace2.silver.sales")

cust_az_silver = spark.table("workspace2.silver.cust_az")

loc_a1_silver = spark.table("workspace2.silver.loc_a1")

px_cat_silver = spark.table("workspace2.silver.px_cat")
```

# CUSTOMERS

```python
customer_silver.display()
```

## Duplication/Nulls

```python
# Check for NULLs or Duplicates in Primary Key
# Expectation: No Results

customer_silver.groupBy("customer_id").count().filter(col("count") > 1).display()

customer_silver.filter(col("customer_id").isNull()).display()
```

## Unwanted Spaces

```python
# Check the unwanted spaces for the firstname and lastname

customer_silver.filter(
    (col("firstname") != trim(col("firstname"))) 
    | 
    (col("lastname") != trim(col("lastname")))
).display()
```

## Standarization

```python
# Check the standarization for marital status and gender

customer_silver.select("marital_status","gender").distinct().display()
```

# PRODUCTS

```python
products_silver.display()
```

## Duplication/Nulls

```python
# Check for NULLs or Duplicates in Primary Key
# Expectation: No Results

products_silver.groupBy("product_id").count().filter(col("count") > 1).display()

products_silver.filter(col("product_id").isNull()).display()
```

```python
# Check the nulls value in prd_cost

products_silver.filter(col("product_cost").isNull()).display()
```

## Standarization

```python
# Check the standarization for prd_line

products_silver.select("product_line").distinct().display()
```

## Validate date

```python
# Check the validation date for product_start_date and product_end_date

products_silver.filter(col("product_start_date") > col("product_end_date")).display()
```

# SALES

```python
sales_silver.display()
```

## Calculation Validation

```python
# Check the validation of the amount of sales is real

sales_silver.filter(
    (col("sales") != (col("quantity") * abs(col("price"))))
).display()


# Check the the quantity is negatif

sales_silver.filter(col("quantity") < 1).display()

# Check the validation of the price is real

sales_silver.filter(
    (col("price") != (col("sales") / col("quantity")))
).display()
```
