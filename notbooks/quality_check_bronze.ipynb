# INIT

```python
from pyspark.sql.functions import col, trim, abs, current_date
```

```python
## Create tables dataframe  

customer_df = spark.table("workspace2.bronze.customers")

products_df = spark.table("workspace2.bronze.products")

sales_df = spark.table("workspace2.bronze.sales")

cust_az_df = spark.table("workspace2.bronze.cust_az")

loc_a1_df = spark.table("workspace2.bronze.loc_a1")

px_cat_df = spark.table("workspace2.bronze.px_cat")
```

# CUSTOMERS

```python
customer_df.display()
```

## Duplication/Nulls

```python
# Check for NULLs or Duplicates in Primary Key
# Expectation: No Results

customer_df.groupBy("cst_id").count().filter(col("count") > 1).display()

customer_df.filter(col("cst_id").isNull()).display()
```

## Unwanted Spaces

```python
# Check the unwanted spaces for the firstname and lastname

customer_df.filter(
    (col("cst_firstname") != trim(col("cst_firstname"))) 
    | 
    (col("cst_lastname") != trim(col("cst_lastname")))
).display()
```

## Standarization


```python
# Check the standarization for marital status and gender

customer_df.select("cst_marital_status","cst_gndr").distinct().display()
```

# PRODUCTS

```python
products_df.display()
```

## duplication/Nulls

```python
# Check for NULLs or Duplicates in Primary Key
# Expectation: No Results

products_df.groupBy("prd_id").count().filter(col("count") > 1).display()

products_df.filter(col("prd_id").isNull()).display()
```

```python
# Check the nulls value in prd_cost

products_df.filter(col("prd_cost").isNull()).display()
```

## Standarization 

```python
# Check the standarization for prd_line

products_df.select("prd_line").distinct().display()
```

## Validate Date

```python
# Check the validation date for prd_start_dt and prd_end_dt

products_df.filter(col("prd_start_dt") > col("prd_end_dt")).display()
```

# SALES

```python
sales_df.display()
```

## Calculation Validation

```python
# Check the validation of the amount of sales is real

sales_df.filter(
    (col("sls_sales") != (col("sls_quantity") * abs(col("sls_price"))))
).display()


# Check the the quantity is negatif

sales_df.filter(col("sls_quantity") < 1).display()

# Check the validation of the price is real

sales_df.filter(
    (col("sls_price") != (col("sls_sales") / col("sls_quantity")))
).display()
```
