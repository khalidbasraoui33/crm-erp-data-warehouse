# DIM_CUSTOMERS

```python
%sql
-- Check for Uniqueness of Customer Key in gold.dim_customers
-- Expectation: No results

SELECT
    customer_number,
    COUNT(*) AS duplicate_count
FROM workspace2.gold.dim_customers
GROUP BY customer_number
HAVING COUNT(*) > 1;
```

# DIM_PRODUCTS

```python
%sql
-- Check for Uniqueness of Product Key in gold.dim_products
-- Expectation: No results


SELECT
    product_number,
    COUNT(*) AS duplicate_count
FROM workspace2.gold.dim_products
GROUP BY product_number
HAVING COUNT(*) > 1;
```

# DATA MODEL

```python
%sql
-- Check the data model connectivity between fact and dimensions


SELECT *
FROM workspace2.gold.fact_sales f
LEFT JOIN workspace2.gold.dim_customers c
ON c.customer_number = f.customer_number
LEFT JOIN workspace2.gold.dim_products p
ON p.product_number = f.product_number
```
